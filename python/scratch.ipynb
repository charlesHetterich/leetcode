```python
from dataclasses import dataclass
from typing import Optional


@dataclass
class Node:
    value: int
    left: Optional["Node"] = None
    right: Optional["Node"] = None

    def __init__(self, v: int):
        self.value = v
```

```python
root = Node(10)
root.left = Node(5)
root.right = Node(15)
root.left.left = Node(-2)
```
