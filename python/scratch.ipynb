```python
from serialize_deserialize_bt import TreeNode, Codec
```

```python
root = TreeNode(1)
root.left = TreeNode(-2)
root.right = TreeNode(3)
root.right.left = TreeNode(4)
root.right.right = TreeNode(5)

Codec().serialize(root)
```

```python
r2 = Codec().deserialize(Codec().serialize(root))

print(r2.val)
print(r2.left.val)
print(r2.right.val)
print(r2.right.left.val)
print(r2.right.right.val)
```
