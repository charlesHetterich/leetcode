```python
import math
def disp_x(x: int) -> str:
    if x == None:
        return "None"
    return str(round(x, 2))

def analyize(seq: list[int]):
    print("\t".join(map(lambda x: str(x), seq)))
    print('\n')

    difs = []
    difs2 = []
    mults = []
    mults2 = []
    log_base = []
    log_base2 = []
    for i in range(1, len(seq)):

        difs.append(seq[i] - seq[i-1])

        try:
            mults.append(seq[i] / seq[i-1])
        except:
            mults.append(None)    
        # try:
        #     log_base.append(math.log(seq[i]))
        # except:
        #     log_base.append(None)

        if i >= 2:
            difs2.append(seq[i] - seq[i-2])
            try:
                mults2.append(seq[i] / seq[i-2])
            except:
                mults2.append(None)

    print("dif1\t" + "\t".join(map(lambda x: disp_x(x), difs)))
    print("dif2\t\t" + "\t".join(map(lambda x: disp_x(x), difs2)))
    print('\n')

    print("mult\t" + "\t".join(map(lambda x: disp_x(x), mults)))
    print("mult2\t\t" + "\t".join(map(lambda x: disp_x(x), mults2)))
```
