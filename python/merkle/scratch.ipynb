```python
from merkle_tree import hash, encode, key_from_bytes, TrieBase

data = [[1, 2], ["22", b"asdfasdfa"], [5, 6], [7, 8]]
hash(data).hexdigest()
```

```python
encoded_data = encode("oye")
k16 = key_from_bytes(encoded_data)
k4 = key_from_bytes(encoded_data, TrieBase.BASE4)
k2 = key_from_bytes(encoded_data, TrieBase.BASE2)

for kk in [k16, k4, k2]:
    print(f"{len(kk)}\t {kk}")
```

```python
cc = encode(True)
for i, b in enumerate(cc):
    l = []
    for i in range(7, -1, -1):
        l.append(b >> i & 1)
    print(f"{b}\t {len(l)}\t {l}")

for i, b in enumerate(cc):
    l = []
    for i in range(6, -1, -2):
        l.append(b >> i & 0b11)
    print(f"{b}\t {len(l)}\t {l}")

print("\t\t", [cc.hex()])
```
